```python
import os
from crewai import Agent, Task, Crew, Process
from langchain_groq import ChatGroq
import numpy as np

np.float_ = np.float64
os.environ["GROQ_API_KEY"] = "your-groq-api-key"

llm = ChatGroq(
    temperature=0.3,
    model_name="meta-llama/llama-4-scout-17b-16e-instruct",  # or any preferred Groq model
)

# Agent 1: Property Data Researcher
data_agent = Agent(
    llm=llm,
    role="Property Data Researcher",
    goal="Gather relevant details of USA residential properties like price, rent, taxes, and location details",
    backstory="An expert in sourcing real estate market data from credible sources such as Zillow, Redfin, and Census data.",
    allow_delegation=False,
    verbose=True
)

# Agent 2: Financial Analyst
finance_agent = Agent(
    llm=llm,
    role="Financial Analyst",
    goal="Evaluate ROI, cap rate, and cash flow of each property",
    backstory="A seasoned financial advisor specializing in real estate investment metrics and forecasting returns.",
    allow_delegation=False,
    verbose=True
)

# Agent 3: Market Trends Analyst
market_agent = Agent(
    llm=llm,
    role="Market Trends Analyst",
    goal="Analyze economic and real estate trends impacting property investments across different states in the USA",
    backstory="A macroeconomic analyst who evaluates housing demand, migration trends, and employment impact on property prices.",
    allow_delegation=False,
    verbose=True
)

# Agent 4: Investment Recommendation Agent
recommender_agent = Agent(
    llm=llm,
    role="Investment Advisor",
    goal="Based on ROI and trends, recommend top 3 cities and properties for investment",
    backstory="A senior investment strategist with decades of experience identifying high-potential real estate markets.",
    allow_delegation=False,
    verbose=True
)

report_agent = Agent(
    llm=llm,
    role="Report Generator",
    goal="Create a professional investor report with property analysis, ROI tables, and recommendations",
    backstory="A business writer responsible for compiling detailed reports for real estate investors and clients.",
    allow_delegation=False,
    verbose=True
)

tasks = [
    Task(
        description="Collect property data like location, price, rental income, property tax, and area-wise crime rates for 10 US cities.",
        expected_output="Structured data for 10 cities with property listings and financial data.",
        agent=data_agent
    ),
    Task(
        description="Evaluate the ROI, cap rate, cash-on-cash return for each property listed.",
        expected_output="Financial analysis table comparing properties.",
        agent=finance_agent
    ),
    Task(
        description="Identify macro and micro real estate trends affecting the top 10 cities in the US.",
        expected_output="Trend report summarizing population, job growth, migration, and price appreciation.",
        agent=market_agent
    ),
    Task(
        description="Select and recommend the top 3 cities and best properties based on ROI and trends.",
        expected_output="A shortlist of 3 investment-worthy cities and properties with reasoning.",
        agent=recommender_agent
    ),
    Task(
        description="Compile a clean PDF-style summary report for the investor with visuals, tables, and recommendations.",
        expected_output="Final report in structured format with summaries and charts.",
        agent=report_agent
    ),
]

real_estate_crew = Crew(
    agents=[data_agent, finance_agent, market_agent, recommender_agent, report_agent],
    tasks=tasks,
    process=Process.sequential,
    verbose=True
)

if __name__ == "__main__":
    result = real_estate_crew.kickoff()
    print("Real Estate Investment Report:", result)
```
